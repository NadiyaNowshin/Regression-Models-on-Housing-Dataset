```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

1. Load the Dataset

```python
# Load the dataset

data_url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/housing.data"

columns = ['CRIM', 'ZN', 'INDUS', 'CHAS', 'NOX', 'RM', 'AGE', 'DIS', 'RAD', 'TAX', 'PTRATIO', 'B', 'LSTAT', 'MEDV']

df = pd.read_csv(data_url, sep=r'\s+', names = columns)
```

```python
# Check the dataset
print("Data set shape:" , df.shape)
print(df.head())
```

2. Split the Dataset

```python
# Separate features (X) and target (y)
X = df.drop('MEDV', axis = 1)
y = df['MEDV']
```

```python
# Split into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.20, random_state = 42)
```

```python
print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

3. Train Models

```python
# Create the models
linear_model = LinearRegression()
ridge_model = Ridge()
lasso_model = Lasso()

# Train the models
linear_model.fit(X_train, y_train)
ridge_model.fit(X_train, y_train)
lasso_model.fit(X_train, y_train)

print("All three models have been trained.")
```

4. Evaluate Models

```python
# Make predictions
liner_pred = linear_model.predict(X_test)
ridge_pred = ridge_model.predict(X_test)
lasso_pred = lasso_model.predict(X_test)
```

```python
# Linear Regression metrics
linear_mse = mean_squared_error(y_test, liner_pred)
linear_rmse = np.sqrt(linear_mse)
linear_r2 = r2_score(y_test, liner_pred)
linear_mae = mean_absolute_error(y_test, liner_pred)

# print Linear Regression metrics
print("Linear Regression")
print("MSE:", linear_mse)
print("RMSE:", linear_rmse)
print("R²:", linear_r2)
print("MAE:", linear_mae)
```

```python
# Ridge Regression metrics
ridge_mse = mean_squared_error(y_test, ridge_pred)
ridge_rmse = np.sqrt(ridge_mse)
ridge_r2 = r2_score(y_test, ridge_pred)
ridge_mae = mean_absolute_error(y_test, ridge_pred)

# print Ridge Regression metrics
print("Ridge Regression")
print("MSE:", ridge_mse)
print("RMSE:", ridge_rmse)
print("R²:", ridge_r2)
print("MAE:", ridge_mae)
```

```python
# Lasso Regression metrics
lasso_mse = mean_squared_error(y_test, lasso_pred)
lasso_rmse = np.sqrt(lasso_mse)
lasso_r2 = r2_score(y_test, lasso_pred)
lasso_mae = mean_absolute_error(y_test, lasso_pred)

# print Lasso Regression metrics
print("Lasso Regression")
print("MSE:", lasso_mse)
print("RMSE:", lasso_rmse)
print("R²:", lasso_r2)
print("MAE:", lasso_mae)
```

5. Compare Results

```python
# Make a comparison table

results = pd.DataFrame({
    'Model': [
        'Linear Regression',
        'Ridge Regession',
        'Lasso Regression'
    ],

    'MSE':[
        linear_mse,
        ridge_mse,
        lasso_mse
    ],

    'RMSE':[
        linear_rmse,
        ridge_rmse,
        lasso_rmse
    ],

    'R²':[
        linear_r2,
        ridge_r2,
        lasso_r2
    ],

    'MAE':[
        linear_mae,
        ridge_mae,
        lasso_mae
    ]
})

print(results.round(4))
```
